## Fold Experiment LLM

```python
import jsonlines, collections
actions = {}
with jsonlines.open("../results/fold_change_llm_bg80.0_episodes.jsonl") as f:
    for ep in f:
        steps = ep["steps"]
        epnum = ep["episode_summary"]["episode"]
        actions[epnum] = []
        for step in steps:
            actions[epnum].append(step["action"])
for epnum, action in actions.items():
    print(epnum, collections.Counter(action))
```
